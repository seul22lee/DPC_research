```python
import pickle

file_path = "/home/ftk3187/github/DPC_research/01_Toy/ME341_Project_DPC/all_policy_losses_randomdata_hiddendimensions.pkl"

with open(file_path, "rb") as f:
    results = pickle.load(f)

    
epochs_to_check = [100, 500, 800]

for setting_name, data in results.items():
    print(f"\n=== {setting_name} ===")
    
    train_loss = data["train_loss"]
    val_loss = data["val_loss"]
    
    for ep in epochs_to_check:
        idx = ep - 1  # index 변환
        
        if idx < len(train_loss):  # early stopping 대비
            print(f"Epoch {ep}: "
                  f"Train Loss = {train_loss[idx]:.6f}, "
                  f"Val Loss = {val_loss[idx]:.6f}")
        else:
            print(f"Epoch {ep}: Not available (stopped early)")
```

```python
import pickle
import numpy as np
import matplotlib.pyplot as plt

# =========================
# 1. Load data
# =========================
file_path = "/home/ftk3187/github/DPC_research/01_Toy/02_draft/all_policy_losses_case3.pkl"

with open(file_path, "rb") as f:
    results = pickle.load(f)

# =========================
# 2. 설정
# =========================
layer_list = [3, 5, 7, 9]
hidden_dim_list = [128, 256, 512]

color_map = {
    128: "green",
    256: "purple",
    512: "black",
}

linestyles = ['-', '--', '-.']

# =========================
# 3. subplot (3행 4열)
# =========================
fig, axes = plt.subplots(3, 4, figsize=(20, 12), sharex=True, sharey='row')

# =========================
# 4. plotting
# =========================
for col, n_layers in enumerate(layer_list):
    ax_loss = axes[0, col]
    ax_const = axes[1, col]
    ax_track = axes[2, col]

    for j, hidden_dim in enumerate(hidden_dim_list):
        key = f"{n_layers}L_{hidden_dim}H"
        if key not in results:
            continue

        color = color_map[hidden_dim]
        style = linestyles[j]

        val_loss = results[key]["val_loss"]
        val_constraint = results[key]["val_constraint_violation"]
        val_tracking = results[key]["val_tracking"]

        # Loss
        ax_loss.plot(val_loss, color=color, linestyle=style, linewidth=2)

        # Constraint
        ax_const.plot(val_constraint, color=color, linestyle=style, linewidth=2)

        # Tracking
        ax_track.plot(val_tracking, color=color, linestyle=style, linewidth=2)

    ax_loss.set_title(f"{n_layers} Layers")
    ax_loss.set_yscale("log")
    ax_loss.grid(True)

    ax_const.grid(True)
    ax_track.grid(True)

    if col == 0:
        ax_loss.set_ylabel("Validation Loss (log scale)")
        ax_const.set_ylabel("Constraint Violation")
        ax_track.set_ylabel("Tracking Loss")

# =========================
# 5. legend (한 번만)
# =========================
axes[0, -1].legend(
    [f"{hd}H" for hd in hidden_dim_list],
    title="Hidden Dim",
    fontsize=9
)

# =========================
# 6. x-label
# =========================
for ax in axes[2]:
    ax.set_xlabel("Epoch")

plt.tight_layout()
plt.show()
```
