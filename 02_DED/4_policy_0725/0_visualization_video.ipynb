```python
import imageio
from pathlib import Path

# 이미지 폴더 경로
img_dir = Path("/home/ftk3187/github/DPC_research/02_DED/4_policy_0725/plots")

# 파일 리스트 정렬
files = sorted(img_dir.glob("mpc_step_*_epi_alea.png"))

# 출력 비디오 파일
output_path = img_dir / "mpc_rollout_both.mp4"

fps = 10  # 프레임 속도

writer = imageio.get_writer(output_path, fps=fps)

for f in files:
    img = imageio.imread(f)
    writer.append_data(img)

writer.close()

print(f"✅ MP4 saved to: {output_path}")
```
